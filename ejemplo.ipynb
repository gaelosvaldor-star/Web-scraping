```python

!pip install -q playwright
!playwright install --with-deps chromium

from pathlib import Path
from playwright.async_api import async_playwright

URL = "https://www.economia-sniim.gob.mx/nuevo/"
CAPTURA = Path("sniim.png")

async def main():
    async with async_playwright() as p:

        navegador = await p.chromium.launch(headless=True)
        pagina = await navegador.new_page(viewport={"width": 1280, "height": 800})

        await pagina.goto(URL)

        print("Título:", await pagina.title())
        print("URL:", pagina.url)

        await pagina.screenshot(path=str(CAPTURA), full_page=True)
        print("Captura guardada en:", CAPTURA.resolve())

        await navegador.close()

# En entornos interactivos con asyncio integrado (como Colab), ejecutamos directamente con await
await main()
```
